```python
# Student Data Management Using Dictionary, Tuple and List
student_list = [
    ("Prayusha", "CSE-AIDS", 85),
    ("Dia", "CSE", 78),
    ("Rohan", "ENTC", 90)
]
students = {
    101: student_list[0],
    102: student_list[1],
    103: student_list[2]
}
# 1. Add a new student record
students[104] = ("Ananya", "CSE-AIDS", 88)
# 2. Delete an existing student record
del students[102]
# 3. Update the details of a student
students[103] = ("Rohan", "CSE", 95)
# 4. Display the final student records
print("Final Student Records:")
for roll_no, details in students.items():
    print("Roll Number:", roll_no)
    print("Name:", details[0])
    print("Branch:", details[1])
    print("Marks:", details[2])
```
