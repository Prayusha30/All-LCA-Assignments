```python
# Grocery Price Tracker
prices = [120, 650, 250, 450, 800, 300]
# Remove items priced above 500
prices = [p for p in prices if p <= 500]
# Insert a new item price at position 2
prices.insert(2, 350)
# Sort in ascending order
prices.sort()
# Print updated list and total cost
print("Updated list:", prices)
print("Total cost:", sum(prices))
```

```python
# Travel Itenary
cities = ["Pune", "Mumbai", "Delhi", "Jaipur"]
# Add 2 cities at the end
cities.append("Goa")
cities.append("Bengaluru")
# Remove one city by name
cities.remove("Delhi")
# Convert list into tuple
cities = tuple(cities)
# Access first, last and middle city
print("First city:", cities[0])
print("Last city:", cities[-1])
print("Middle city:", cities[len(cities)//2])
```

```python
# Sports Tournament
scores = [78, 92, 65, 88, 74]
# Sort in descending order
scores.sort(reverse=True)
# Convert list into tuple
scores = tuple(scores)
# Unpack tuple
a, b, c, d, e = scores
print("Highest score:", a)
print("Lowest score:", e)
print("Average score:", sum(scores) / len(scores))
```

```python
# Student Records Management System
students = ["Amit", "Neha", "Rahul", "Sneha", "Rahul"]
# 2. Append Vikas
students.append("Vikas")
# 3. Insert Priya at position 2
students.insert(2, "Priya")
# 4. Count Rahul
print("Rahul appears:", students.count("Rahul"), "times")
# 5. Find index of Neha
print("Index of Neha:", students.index("Neha"))
# 6. Remove first occurrence of Rahul
students.remove("Rahul")
# 7. Pop last student
removed_student = students.pop()
print("Removed student:", removed_student)
# 8. Copy list
backup_students = students.copy()
# 9. Extend backup_students
backup_students.extend(["Karan", "Meena"])
# 10. Sort alphabetically
backup_students.sort()
# 11. Reverse list
backup_students.reverse()
print("Students:", students)
print("Backup students:", backup_students)
```

```python
from datetime import date
d1 = int(input("Enter day of first date: "))
m1 = int(input("Enter month of first date: "))
y1 = int(input("Enter year of first date: "))
d2 = int(input("Enter day of second date: "))
m2 = int(input("Enter month of second date: "))
y2 = int(input("Enter year of second date: "))
date1 = date(y1, m1, d1)
date2 = date(y2, m2, d2)
days = abs((date2 - date1).days)
print("Number of days:", days)
```

```python
#prime number check and factor display
n = int(input("Enter a number: "))
factors = []
for i in range(1, n + 1):
    if n % i == 0:
        factors.append(i)
print("Factors:", factors)
if len(factors) == 2:
    print("Prime number")
else:
    print("Not a prime number")
```

```python
#second largest number without sort
numbers = []
for i in range(5):
    n = int(input("Enter number: "))
    numbers.append(n)
largest = numbers[0]
for n in numbers:
    if n > largest:
        largest = n
second_largest = None
for n in numbers:
    if n != largest:
        if second_largest is None or n > second_largest:
            second_largest = n
print("Largest:", largest)
print("Second largest:", second_largest)
```

```python
#sum of digits
n = int(input("Enter a number: "))
sum_digits = 0
while n > 0:
    digit = n % 10
    sum_digits += digit
    n = n // 10
print("Sum of digits:", sum_digits)
```

```python
#reverse number using for loop
n = int(input("Enter an integer: "))
reverse = 0
for i in range(len(str(n))):
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10
print("Reversed number:", reverse)
```

```python
#common element in 2 lists
list1 = [1, 2, 3, 4, 5]
list2 = [6, 7, 3, 8, 9]
common = False
for i in list1:
    for j in list2:
        if i == j:
            common = True
            break
    if common:
        break
if common:
    print("Lists have at least one common element")
else:
    print("Lists have no common element")
```
